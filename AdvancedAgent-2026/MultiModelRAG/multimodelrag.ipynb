```python
import os
from dotenv import load_dotenv
# 1. Change the import to use the Chat model wrapper
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

# 2. Initialize the chat model class
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

# This will now complete in under 1 second
res = llm.invoke("hi, how are you")
print(res.content) # Use .content to extract just the text string cleanly
```
