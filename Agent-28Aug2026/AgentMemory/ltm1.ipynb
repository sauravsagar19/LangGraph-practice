```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from dotenv import load_dotenv
import os
import pprint

load_dotenv()
embedding_model=OpenAIEmbeddings(
    model="text-embedding-3-small",
    api_key=os.getenv("OPENAI_API_KEY")
)
```

### Normal Creation, searching, get in longterm memory

```python
from langgraph.store.memory import InMemoryStore
store=InMemoryStore()

namespace=("ltm1","p_01") # It will create long term db . a folder of ltm1/p_01.
```

```python
res=store.get(namespace,1)
print(res) # because nothing is put there.
```

```python
store.put(namespace,1,"My name is saurav sagar")
store.put(namespace,2,"I love python and generative ai")
store.put(namespace,3,"I went to lonavala, mahabaleswar onces.")
store.put(namespace,4,{"data":"My home town is purnia bihar"})
```

```python
store.get(namespace,1) # It will return "My name is saurav sagar"
```

```python
# Search : it usually searches in namespace.
#SEARCH
results = store.search(
    namespace, #Required (Positional) 
    query=None,           # Optional
    filter=None,          # Optional
    limit=10,             # Optional
    offset=0              # Optional
)
pprint.pprint(results)
```

```python
results=store.search(
    namespace,
    query="what is my fav subject",           # Optional
)
pprint.pprint(results)


#Note: it will still fetch the all data from namespace as we have not set it semantically.
```

## Semantic search in longterm memroy (VVVI)

```python
store=InMemoryStore(
    index={"dims": 1536,"embed": embedding_model}
)
namespace2=("ltm1","p_02")
```

```python
store.put(namespace2,1,"My name is saurav sagar")
store.put(namespace2,2,"I love python and generative ai")
store.put(namespace2,3,"I went to lonavala, mahabaleswar onces.")
store.put(namespace2,4,{"data":"My home town is purnia bihar"})
```

```python
results=store.search(namespace2,query="what is my name", limit=3)
for r in results:
    print(r.value)
    print(r.score)
    print(r.namespace)
    print("------------------")
```
